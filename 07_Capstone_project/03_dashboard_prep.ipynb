```python
import pandas as pd
from pathlib import Path
PROJECT_DIR = Path.cwd()
df = pd.read_csv(PROJECT_DIR / 'data' / 'processed' / 'cleaned_data.csv')
df = df.dropna(subset=['cuisines']).copy()
df['cuisine_individual'] = df['cuisines'].str.split(',')
df_exploded = df.explode('cuisine_individual')
df_exploded['cuisine_individual'] = df_exploded['cuisine_individual'].str.strip().str.title()
df_exploded = df_exploded[df_exploded['cuisine_individual'].ne('')]
output = PROJECT_DIR / 'data' / 'processed' / 'cleaned_data_cuisine_exploded.csv'
output.parent.mkdir(parents=True, exist_ok=True)
df_exploded.to_csv(output, index=False)
print('Saved', output)
```
